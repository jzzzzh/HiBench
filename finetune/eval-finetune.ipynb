```python
import re
import json
```

```python
with open('/home/zhuohang/disk/HiBench/unsloth_finetune/241216_output.json', 'r') as f:
    results = json.load(f)
results
```

```python
def extract_string_from_dict(string: str, key: str) -> str:
    # Regex to match key-value pairs and capture the entire value associated with the key
    pattern = (
        r"'?" + re.escape(key) + r"'?\s*[:=]\s*"  # Match the key with optional quotes and : or =
        r"(\{.*?\}|[^\{\}]+)"  # Capture either a full dictionary-like structure or a complex string value
    )
    match = re.search(pattern, string)
    if not match:
        return None

    # Return the captured value, cleaned for extra whitespace
    value = match.group(1)
    return value.strip() if value else None

def string_match(source: str, target: str) -> bool:
    source = source.lower()
    target = target.lower()
    source = re.sub(r'\s+', ' ', source)
    target = re.sub(r'\s+', ' ', target)
    source = re.sub(r'[^a-zA-Z0-9]', '', source)
    target = re.sub(r'[^a-zA-Z0-9]', '', target)
    if target is None or target == '':
        return False
    if target == source or target in source or source in target:
        return True
    else:
        return False
```

```python
total_questions = 0
correct_answers = 0

for question in results:
    total_questions += 1
    ans = str(question.get("output"))
    ans = extract_string_from_dict(ans, 'answer')
    ans = str(ans)
    ref_ans = str(question.get("true_ans"))
    ref_ans = extract_string_from_dict(ref_ans, 'answer')
    ref_ans = str(ref_ans)
    print(ans)
    print(ref_ans)
    if string_match(ans, ref_ans):
        correct_answers += 1

accuracy = correct_answers / total_questions if total_questions > 0 else 0

print(f"Total Questions: {total_questions}\n")
print(f"Correct Answers: {correct_answers}\n")
print(f"Accuracy: {accuracy:.2%}\n")
```
