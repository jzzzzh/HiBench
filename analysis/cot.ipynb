# Fine Granularity

```python
binary_ret = dict()
binary_tree = pd.read_csv('format-results/binary-tree.csv')
binary_tree = binary_tree[binary_tree['ExampleType'] == 'ZeroShot']
binary_tree = binary_tree[binary_tree['ModelName'].isin(['Qwen2.5-32B-Instruct', 'Qwen2.5-7B-Instruct'])].replace('Qwen2.5-32B-Instruct', 'QwQ-32B')
binary_ret['Qwen-7B'] = binary_tree[binary_tree['ModelName'] == 'Qwen2.5-7B-Instruct'].groupby('SubTask')['Accuracy'].mean()
binary_ret['QwQ-32B'] = binary_tree[binary_tree['ModelName'] == 'QwQ-32B'].groupby('SubTask')['Accuracy'].mean()

binary_tree_cot = pd.read_csv('./format-results/binary-tree-CoT.csv')
binary_ret['Qwen-7B-CoT'] = binary_tree_cot[binary_tree_cot['ModelName'] == 'Qwen2.5-7B-Instruct'].groupby('SubTask')['Accuracy'].mean()
binary_ret['QwQ-32B-CoT'] = binary_tree_cot[binary_tree_cot['ModelName'] == 'QwQ-32B'].groupby('SubTask')['Accuracy'].mean()
pd.DataFrame(binary_ret)
```

```python
multiple_ret = dict()
multiple_tree = pd.read_csv('format-results/multiple-tree.csv')
multiple_tree = multiple_tree[multiple_tree['ExampleType'] == 'ZeroShot']
multiple_tree = multiple_tree[multiple_tree['ModelName'].isin(['Qwen2.5-32B-Instruct', 'Qwen2.5-7B-Instruct'])].replace('Qwen2.5-32B-Instruct', 'QwQ-32B')
multiple_ret['Qwen-7B'] = multiple_tree[multiple_tree['ModelName'] == 'Qwen2.5-7B-Instruct'].groupby('SubTask')['Accuracy'].mean()
multiple_ret['QwQ-32B'] = multiple_tree[multiple_tree['ModelName'] == 'QwQ-32B'].groupby('SubTask')['Accuracy'].mean()

multiple_tree_cot = pd.read_csv('./format-results/multiple-tree-CoT.csv')
multiple_ret['Qwen-7B-CoT'] = multiple_tree_cot[multiple_tree_cot['ModelName'] == 'Qwen2.5-7B-Instruct'].groupby('SubTask')['Accuracy'].mean()
multiple_ret['QwQ-32B-CoT'] = multiple_tree_cot[multiple_tree_cot['ModelName'] == 'QwQ-32B'].groupby('SubTask')['Accuracy'].mean()
pd.DataFrame(multiple_ret)
```
